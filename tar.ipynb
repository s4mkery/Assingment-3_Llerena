```python
import numpy as np

info = np.genfromtxt("NBA_Player_Stats.tsv", delimiter= "\t", skip_header=1 , dtype = float)
```

### Accuracy Percentage Formula = Made / Attempted * 100

```python
#Field Goal Accuracy

field_arr = info[:,7] / info[:,8] * 100
print(field_arr)

#Three Point Accuracy
threeP_arr = info[:,9] / info[:,10] * 100
print(threeP_arr)

#Free Throw Accuracy
free_arr = info[:,11] / info[:,12] * 100
print(free_arr)
```
