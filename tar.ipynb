```python
import numpy as np

info = np.genfromtxt("NBA_Player_Stats.tsv", delimiter= "\t", skip_header=1 , dtype = str)
```

### Accuracy Percentage Formula = Made / Attempted * 100

```python

players_season = info[:, [1,3]]
#Field Goal Accuracy
fg_Made = info[:,7].astype(float)
fg_Attempted = info[:,8].astype(float)
field_arr = np.where(fg_Attempted>0 , (fg_Made / fg_Attempted) * 100, 0)
print(np.column_stack((players_season,field_arr)))

#Three Point Accuracy
threeP_Made = info[:,9].astype(float)
threeP_Attempted = info[:,10].astype(float)
threeP_arr = np.where(threeP_Attempted > 0, (threeP_Made / threeP_Attempted) * 100, 0)
print(np.column_stack((players_season, threeP_arr)))

#Free Throw Accuracy
freeT_Made = info[:,11].astype(float)
freeT_Attempted = info[:,12].astype(float)
freeT_arr = np.where(freeT_Attempted > 0,(freeT_Made / freeT_Attempted) * 100, 0)
print(np.column_stack((players_season, freeT_arr)))
```

```python
# B)Average points per minute
minutes_played = info[:,6].astype(float)
points = info[:,21].astype(float)
avg_points = np.where(minutes_played > 0, points / minutes_played, 0)
print(np.column_stack((players_season, avg_points)))
```

```python
#C)Overall shooting accuracy
overall_Shooting = (field_arr + threeP_arr + freeT_arr) / 3
print(np.column_stack((players_season, overall_Shooting)))
```

```python
#D) Average number of blocks and Average number of steals
games_played = info[:,5].astype(float)
blocks = info[:,20].astype(float)
steals = info[:,19].astype(float)

avg_blocks = np.where(games_played > 0, blocks / games_played, 0)
print(np.column_stack((players_season, avg_blocks)))

avg_steals = np.where(games_played > 0, steals / games_played, 0)
print(np.column_stack((players_season, avg_steals)))
```

```python
#2  Top 100

#Top Field goal acurracy
fieldG_idx = np.argsort(-field_arr)[:100]
fieldG_players= players_season[fieldG_idx]
fieldG_values = field_arr[fieldG_idx]

top_fieldG  = np.column_stack((fieldG_players, fieldG_values))

#Top Three points accuracy
threeP_idx = np.argsort(-threeP_arr)[:100]
threeP_players = players_season[threeP_idx]
threeP_values = threeP_arr[threeP_idx]

top_threeP  = np.column_stack((threeP_players, threeP_values))

#Top Free throw accuracy
freeT_idx = np.argsort(-freeT_arr)[:100]
freeT_players = players_season[freeT_idx]
freeT_values = freeT_arr[freeT_idx]

top_freeT = np.column_stack((freeT_players, freeT_values))

#Top Average Points accuracy
points_idx = np.argsort(-avg_points)[:100]
points_players= players_season[points_idx]
points_values = avg_points[points_idx]

top_avg_points = np.column_stack((points_players, points_values))

#Top Overall Shooting accuracy
overall_idx = np.argsort(-overall_Shooting)[:100]
overall_players = players_season[overall_idx]
overall_values = overall_Shooting[overall_idx]

top_overall_shooting = np.column_stack((overall_players, overall_values))

#Top Average Blocks
blocks_idx = np.argsort(-avg_blocks)[:100]
blocks_players = players_season[blocks_idx]
blocks_values = avg_blocks[blocks_idx]

top_blocks = np.column_stack((blocks_players, blocks_values))


#Top Steals
steals_idx = np.argsort(-avg_steals)[:100]
steals_players= players_season[steals_idx]
steals_values = avg_steals[steals_idx]

top_steals  = np.column_stack((steals_players, steals_values))



```
